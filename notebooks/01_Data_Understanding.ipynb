# CredIntel – Credit Intelligence & Risk Analytics


**Data Understanding and Exploration**


**Project Overview:**

This project aims to analyze loan application data to understand the key factors influencing loan approval decisions and customer credit risk. The objective is to identify patterns in applicant profiles, financial strength, credit scores, and asset ownership that impact lending decisions.

The insights generated from this analysis can help financial institutions improve risk assessment, optimize loan approval strategies, and support data-driven lending decisions.


**Business Problem:**

Financial institutions must evaluate thousands of loan applications while minimizing the risk of default. The challenge is to identify high-risk applicants and approve loans for financially reliable customers.


**This project focuses on:**

Understanding customer financial profiles.

Analyzing loan approval and rejection patterns.

Evaluating the impact of CIBIL scores.

Assessing income and asset strength.

Building a foundation for credit risk prediction and loan approval modeling

**Dataset Information:**

The dataset contains loan application records with applicant demographics, financial information, asset details, credit scores, and loan approval status.

**Key attributes include:**

Annual Income

Loan Amount

Loan Term

CIBIL Score

Education Level

Employment Status

Residential Assets

Commercial Assets

Luxury Assets

Bank Assets

Loan Status
